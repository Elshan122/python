## 1. Что исследуем простыми словами

Вопрос: у большинства экономик с ростом реального ВВП на душу с 2010 по 2019 год улучшились хотя бы три из четырёх социальных показателей? Мы не спрашиваем, стали ли абсолютно все жители богаче. ВВП показывает выпуск на человека, а не личный доход каждого. Сначала определяем, где ВВП вырос, затем внутри этой группы считаем социальные улучшения. Рост продолжительности жизни и доступа к электричеству — улучшения; снижение смертности и безработицы — тоже улучшения. Равенство не засчитывается. Не менее трёх означает три или четыре. Большинство означает строго больше половины.

## 2. Какие числа нужно понимать

217 экономик находятся в файле; на каждую две строки за два года, поэтому 434 строки данных плюс заголовок. 43 экономики исключены из расчётов из-за хотя бы одного пропуска. Остаются 174 с полными десятью значениями: пять показателей × два года. У 150 из них ВВП на душу вырос. У 131 из этих 150 есть минимум три социальных улучшения. Поэтому доля равна 131 / 150 × 100 = 87,33%. Делить на 174 нельзя: вопрос задан именно об экономиках с ростом ВВП. Среди 150 распределение улучшений: 0 — 0 экономик, 1 — 1, 2 — 18, 3 — 77, 4 — 54. Проверка: 77 + 54 = 131; 1 + 18 = 19; всего 150.

## 3. Откуда взялись данные и почему пять показателей

## 4. Что происходит с одной строкой CSV

Учебный пример, не наблюдение Всемирного банка: AAA;Example;2010;1000;70;20;8;90. После strip().split(";") получаем восемь текстовых элементов. parts[0] — AAA, parts[1] — Example, parts[2] — 2010, parts[3] — 1000. Цикл по i=0…4 берёт parts[i+3] и связывает его с indicators[i]. После преобразования результат читается как data["AAA"][2010]["gdp_pc"] = 1000.0. Вторая строка для 2019 добавляет ещё один год внутрь того же кода. Если поле выглядит как ;;, оно остаётся пустой строкой. Мы не превращаем неизвестную безработицу в 0%: это означало бы совершенно другое наблюдение.

## 5. Как устроены словари и списки

data: код → год → показатель → значение. names: код → название. complete: список кодов с полными данными. excluded: код → объяснение пропусков. stats: год → показатель → [среднее, медиана, минимум, максимум, размах]. changes: код → показатель → разность 2019−2010. growth: код → процент прироста ВВП. extremes: показатель → [минимум, названия минимума, максимум, названия максимума]. improvements: код → число улучшений. growing и successful — списки кодов, их длины дают знаменатель и числитель. distribution: число улучшений → число экономик. rating и top10 — списки кортежей. report — список текстовых строк. Имя row — временное сокращение: в разных местах оно обозначает разные списки или словари; смотрите на ближайшее присваивание.

## 6. Пять статистик на маленьком примере

Учебный список [2, 4, 9, 13]: среднее = (2+4+9+13)/4 = 7; медиана = (4+9)/2 = 6,5; минимум = 2; максимум = 13; размах = 13−2 = 11. Среднее использует все значения, медиана — положение центра упорядоченного ряда, минимум и максимум — границы, размах — расстояние между ними. Это пять разных видов статистики, хотя размах вычисляется из двух других. Для пяти полей и двух лет получается 50 числовых результатов, но пять видов статистики. sorted возвращает новый отсортированный список. Индексы начинаются с нуля: для 174 значений центральные позиции 87 и 88 имеют индексы 86 и 87. Мы не взвешиваем экономики по населению; среднее по странам не равно среднему по людям.

## 7. Проценты, процентные пункты и направление

Учебный пример: ВВП вырос с 1000 до 1200. Абсолютная разница = 200 долларов, относительный прирост = (1200/1000−1)×100 = 20%. Это прирост за девять лет, не 20% ежегодно. Безработица упала с 8% до 6%: разница −2 процентных пункта; относительное изменение −25%, но его в проекте для безработицы не считаем. В словаре directions для безработицы стоит −1: (−2)×(−1)=2>0, поэтому это улучшение. При электричестве 100% в обоих годах разница 0 — сохранение хорошего уровня не новое улучшение. Минимальная разность не всегда худшая: самое большое снижение смертности является минимальной разностью и желательным изменением.

## 8. Почему запускаем сверху вниз

Первая ячейка создаёт data, names и complete. Вторая использует их и создаёт stats. Третья создаёт changes, growth и extremes. Четвёртая использует changes и growth для answer и share. ## 9. Ячейка 1: Чтение CSV и отбор полной выборки

### Как объяснить устно

Открываем файл, делим каждую строку на поля и складываем её в словарь по коду и году. Затем у каждой экономики проверяем десять значений. Если пропусков нет, включаем код в complete, иначе сохраняем причину исключения. Получаем 217 исходных, 174 полных и 43 исключённых экономики.

````python
# Словарь хранит все записи, включая строки с пропусками.
data = {}
names = {}
indicators = ['gdp_pc', 'life', 'infant', 'unemployment', 'electricity']
years = [2010, 2019]
with open('economy_data.csv', 'r', encoding='utf-8') as input_file:
    for line in input_file:
        parts = line.strip().split(';')
        if parts[0] != 'code':
            country = parts[0]
            year = int(parts[2])
            names[country] = parts[1]
            if country not in data:
                data[country] = {}
            data[country][year] = {}
            for i in range(len(indicators)):
                value = parts[i + 3]
                if value != '':
                    value = float(value)
                data[country][year][indicators[i]] = value
print('Экономик в файле:', len(data))
# Для двух лет используем одинаковую выборку с полными наблюдениями.
complete = []
excluded = {}
for country in data:
    missing = []
    for year in years:
        for indicator in indicators:
            if data[country][year][indicator] == '':
                missing.append(str(year) + ':' + indicator)
    if len(missing) == 0:
        complete.append(country)
    else:
        excluded[country] = ', '.join(missing)
print('Полных наблюдений за два года:', len(complete))
print('Исключено из расчётов:', len(excluded))
for country in excluded:
    print(country, names[country], excluded[country], sep=';')
````

## 10. Ячейка 2: Пять статистик в собственной функции

### Как объяснить устно

Для каждой пары год–показатель собираем список из 174 чисел. Передаём список функции describe. Внутри сортируем для медианы, считаем среднее, границы и размах и возвращаем список результатов. Сохраняем его в stats.

````python
# Одна функция используется для каждого показателя и каждого года.
def describe(values):
    ordered = sorted(values)
    count = len(ordered)
    mean = sum(values) / count
    if count % 2 == 1:
        median = ordered[count // 2]
    else:
        median = (ordered[count // 2 - 1] + ordered[count // 2]) / 2
    minimum = min(values)
    maximum = max(values)
    return [mean, median, minimum, maximum, maximum - minimum]

stats = {}
print('Год;Показатель;Среднее;Медиана;Минимум;Максимум;Размах')
for year in years:
    stats[year] = {}
    for indicator in indicators:
        values = []
        for country in complete:
            values.append(data[country][year][indicator])
        row = describe(values)
        stats[year][indicator] = row
        print(year, indicator, round(row[0], 3), round(row[1], 3), round(row[2], 3), round(row[3], 3), round(row[4], 3), sep=';')
````

## 12. Ячейка 4: Проверка исследовательского вопроса

### Как объяснить устно

У каждой экономики обнуляем счётчик и проверяем четыре направления изменения. Сохраняем число улучшений. Если ВВП вырос, включаем экономику в знаменатель; если улучшений минимум три, включаем также в числитель. Находим долю и сравниваем с 50%.

````python
# Для жизни и электричества улучшение — рост; для остальных — снижение.
directions = {'life': 1, 'infant': -1, 'unemployment': -1, 'electricity': 1}
improvements = {}
growing = []
successful = []
distribution = {0: 0, 1: 0, 2: 0, 3: 0, 4: 0}
for country in complete:
    count = 0
    for indicator in directions:
        if changes[country][indicator] * directions[indicator] > 0:
            count += 1
    improvements[country] = count
    if growth[country] > 0:
        growing.append(country)
        distribution[count] += 1
        if count >= 3:
            successful.append(country)
share = len(successful) / len(growing) * 100
if share > 50:
    answer = 'Да'
else:
    answer = 'Нет'
print('Экономик с ростом ВВП на душу:', len(growing))
print('Из них с тремя или четырьмя улучшениями:', len(successful))
print('Доля, %:', round(share, 2))
print('Ответ:', answer)
print('Число улучшений;Число экономик с ростом ВВП')
for count in distribution:
    print(count, distribution[count], sep=';')
print('Экономики с ростом ВВП, но менее чем тремя улучшениями:')
for country in growing:
    if improvements[country] < 3:
        print(names[country], improvements[country], sep=';')
````

## 18. Вопросы преподавателя и короткие ответы

Почему порог 3 из 4? Это явно заданное рабочее правило широкого улучшения, допускающее один показатель без роста; это не официальный индекс. Почему одинаковые страны в обоих годах? Чтобы изменение статистик не отражало смену состава выборки. Почему не заменили пропуски нулями? Ноль — реальное значение, а пропуск — отсутствие информации. Почему 174, а не 217? У 43 экономик нет хотя бы одного необходимого значения. Почему доля 131/150? Вопрос условен на рост ВВП, а растущих экономик 150. Почему не используем импорт? Таково ограничение задания; все расчёты выполнены встроенными средствами. def — импорт? Нет, это наша собственная функция. Почему минус перед growth в рейтинге? Чтобы обычная сортировка по возрастанию дала убывание исходного прироста. Почему не сортируем только названия? Тогда получится алфавитный список, а не рейтинг роста.

## 19. Ещё вопросы о методике

Можно ли утверждать, что уровень жизни измерен полностью? Нет: не учтены неравенство, качество услуг, безопасность и другие стороны. Не дублируются ли показатели здоровья? Частично связаны, поэтому два из четырёх условий относятся к здоровью; это ограничение отмечено. Что с электричеством при 100%? Роста нет, условие не засчитывается, но ещё три показателя могут улучшиться. Все экономики одинаково важны? В этом вопросе считаем долю экономик, поэтому каждая имеет вес 1; для доли населения потребовались бы веса. Почему не нужен csv-модуль? В подготовленном файле разделитель не встречается внутри названий; обычный split работает. Для произвольного CSV с кавычками и разделителями внутри полей такой разбор может быть неверным. Что при пустой полной выборке или отсутствии растущих экономик? Текущий код рассчитан на приложенный проверенный файл; при других данных понадобились бы проверки перед делением. Мы не выдаём его за универсальную программу для любых файлов.
